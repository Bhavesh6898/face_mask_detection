```python
from tensorflow.keras.preprocessing.image import img_to_array
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from tensorflow.keras.models import load_model
import numpy as np
import cv2
```

```python
model = load_model("mobilenetv2_4300_32_100.keras")
```

```python
image_path = "001.jpg"
```

```python
image = cv2.imread(image_path)
```

```python
if image is None:
    raise ValueError(f"Image not found at path: {image_path}")
```

```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
image_resized = cv2.resize(image_rgb, (96, 96))
```

```python
# Preprocess the image as MobileNetV2 expects
image_array = img_to_array(image_resized)
image_array = preprocess_input(image_array)
image_array = np.expand_dims(image_array, axis=0)  # Add batch dimension
```

```python
(mask, withoutMask) = model.predict(image_array)[0]
```

```python
# Determine label and confidence
label = "Mask" if mask > withoutMask else "No Mask"
confidence = max(mask, withoutMask) * 100
color = (0, 255, 0) if label == "Mask" else (0, 0, 255)
```

```python
# Display result on the original image
output = image.copy()
cv2.putText(output, f"{label}: {confidence:.2f}%", (10, 30),
            cv2.FONT_HERSHEY_SIMPLEX, 1.0, color, 2)
cv2.imshow("Prediction", output)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
